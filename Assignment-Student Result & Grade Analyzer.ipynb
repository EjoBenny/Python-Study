```python

print("STUDENT RESULT ANALYZER")

# Function to calculate total
def calculate_total(marks):
    return recursive_sum(marks)


# Function to calculate average
def calculate_average(total, count):
    return total / count
    
# Function to calculate total using recursion
def recursive_sum(marks, index=0):
    if index == len(marks):
        return 0
    return marks[index] + recursive_sum(marks, index + 1)

# Function to calculate grade
def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


# Function to display grade-based pattern
def display_pattern(grade):
    if grade == "A+":
        print("Pattern: *****")
    elif grade == "A":
        print("Pattern: ****")
    elif grade == "B":
        print("Pattern: ***")
    elif grade == "C":
        print("Pattern: **")
    elif grade == "D":
        print("Pattern: *")
    else:
        print("Pattern: X")


# Function to display result
def display_result(student_name, roll_number, course_name,
                   subjects, marks, total, average, grade, result):

    print("")
    print("STUDENT RESULT")


    print(f"Student Name : {student_name}")
    print(f"Roll Number  : {roll_number}")
    print(f"Course Name  : {course_name}")

    print("\nSubject Marks")


    # For loop with range()
    for i in range(5):
        print(f"{subjects[i]:<15}: {marks[i]}")


    print(f"Total Marks   : {total} / 500")
    print(f"Average Mark  : {average:.2f}")
    print(f"Grade         : {grade}")
    print(f"Result        : {result}")

    display_pattern(grade)

    print("=" * 50)


# Lambda function for pass/fail
# Average 50 or above = PASS
check_pass = lambda average: "PASS" if average >= 50 else "FAIL"


# Main program
while True:

    # Student details
    student_name = input("\nEnter Student Name   : ")
    roll_number = int(input("Enter Roll Number    : "))
    course_name = input("Enter Course Name    : ")

    print("\nEnter marks for 5 subjects\n")

    # Simultaneous assignment with type casting
    python, html, css, javascript, django = (
        int(input("Python      : ")),
        int(input("HTML        : ")),
        int(input("CSS         : ")),
        int(input("JavaScript  : ")),
        int(input("Django      : "))
    )

    subjects = ("Python", "HTML", "CSS", "JavaScript", "Django")
    marks = (python, html, css, javascript, django)

    # Calculate total using function
    total = calculate_total(marks)

    # Calculate average using function
    average = calculate_average(total, len(marks))

    # Calculate grade using function
    grade = calculate_grade(average)

    # Pass/Fail using lambda
    result = check_pass(average)

    # Display complete result
    display_result(
        student_name,
        roll_number,
        course_name,
        subjects,
        marks,
        total,
        average,
        grade,
        result
    )

    # Continue until user chooses "no"
    choice = input(
        "\nDo you want to enter another student? (yes/no): "
    ).lower()

    if choice == "no":
        print("\nThank you for using Student Result Analyzer!")
        break
```
