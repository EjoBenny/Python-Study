```python
#  Inverted Pyramid
# * * * * * * * * *
#   * * * * * * *
#     * * * * *
#       * * *
#         *

for i in range(5,0,-1):
    print(" "*(5-i) + "*"*((i*2)-1) )
```

```python
# 1. Square Stars
# * * * * *
# * * * * *
# * * * * *
# * * * * *
# * * * * *

for i in range(4):
    for j in range(4):
        print("*",end="")
    print("")
print(id(j))
```

```python
# 2. Increasing Triangle
# *
# * *
# * * *
# * * * *
# * * * * *
print(j)
print(id(j))

for i in range(5):
    for k in range(0,i+1):
        print("*",end="")
    print("")
```

```python
# 3. Decreasing Triangle
# * * * * *
# * * * *
# * * *
# * *
# *

for i in range(5,0,-1):
    for k in range(i,0,-1):
        print("*",end="")
    print("")
```

```python
for i in range(5):
    for j in range(5 - i):
        print("*", end="")
    print()
```
